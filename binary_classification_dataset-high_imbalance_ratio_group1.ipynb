- TabularSetSSM

Protocol highlights:
- OpenML classification datasets (binary or multiclass)
- Shared `StratifiedKFold(n_splits=10, shuffle=True, random_state=42)` splits across all models
- Metrics per fold: ROC-AUC, Accuracy, F1, Log Loss, ECE
- Aggregation to dataset-level means, then across-dataset means/stds
- Mean Normalized metrics and Wins computation
- Mean time (training + inference)

```python
# Shared evaluation, preprocessing, and model-run helpers
MODEL_ORDER = [
    "tabularsetssm",
    "catboost",
    "xgboost",
    "lightgbm",
    "tabnet",
    "ebm",
]

METRIC_COLS = ["roc_auc", "accuracy", "f1", "log_loss", "ece"]
HIGHER_IS_BETTER = {"roc_auc": True, "accuracy": True, "f1": True, "log_loss": False, "ece": False}
N_ECE_BINS = 15
N_SPLITS = 5
INNER_VAL_RATIO = 0.20
EARLY_STOPPING_PATIENCE = 30

MODEL_PARAMS = {
    "tabularsetssm": {
        "hidden_dim": 64,
        "pe_dim": 8,
        "pe_num_filters": 4,
        "rank_dim": 32,
        "num_blocks": 2,
        "dropout": 0.10,
        "learning_rate": 3e-4,
        "weight_decay": 5e-4,
        "batch_size": 128,
        "max_epochs": 80,
        "val_monitor_start_epoch": 10,
    },
    "catboost": {
        "iterations": 600,
        "learning_rate": 0.05,
        "depth": 6,
        "task_type":"GPU"
    },
    "xgboost": {
        "n_estimators": 1200,
        "learning_rate": 0.03,
        "max_depth": 6,
        "subsample": 0.9,
        "colsample_bytree": 0.9,
    },
    "lightgbm": {
        "n_estimators": 1500,
        "learning_rate": 0.03,
    },
    "tabnet": {
        "learning_rate": 2e-2,
        "max_epochs": 200,
        "batch_size": 1024,
        "virtual_batch_size": 128,
    },
    "ebm": {
        "learning_rate": 0.03,
        "max_rounds": 600,
    },
}


def _p_extras(p: dict, *omit: str) -> dict:
    """Pass through extra keys from MODEL_PARAMS to sklearn-style estimators."""
    banned = set(omit) | {"val_monitor_start_epoch"}
    return {k: v for k, v in p.items() if k not in banned and not str(k).startswith("_")}


def format_dataset_key(meta: dict) -> str:
    return f"{meta['openml_id']}_{meta['name']}"


def timestamp_now() -> str:
    return datetime.now().strftime("%Y%m%d_%H%M%S")


def validation_eval_start_epoch(total_epochs: int) -> int:
    eval_epochs = max(1, int(math.ceil(total_epochs * 0.10)))
    return max(1, total_epochs - eval_epochs + 1)


def expected_calibration_error_toplabel(y_true: np.ndarray, prob: np.ndarray, n_bins: int = 15) -> float:
    y_true = np.asarray(y_true).astype(int)
    prob = np.asarray(prob, dtype=float)
    if prob.ndim == 1:
        prob = np.vstack([1.0 - prob, prob]).T

    conf = prob.max(axis=1)
    pred = prob.argmax(axis=1)
    correct = (pred == y_true).astype(float)

    bins = np.linspace(0.0, 1.0, n_bins + 1)
    bin_ids = np.digitize(conf, bins[1:-1], right=False)

    ece = 0.0
    n = len(y_true)
    for b in range(n_bins):
        mask = bin_ids == b
        if not np.any(mask):
            continue
        bin_conf = conf[mask].mean()
        bin_acc = correct[mask].mean()
        ece += (mask.sum() / n) * abs(bin_acc - bin_conf)
    return float(ece)


def compute_classification_metrics(y_true: np.ndarray, prob: np.ndarray) -> dict:
    y_true = np.asarray(y_true).astype(int)
    prob = np.asarray(prob, dtype=float)

    if prob.ndim == 1:
        prob = np.vstack([1.0 - prob, prob]).T

    prob = np.clip(prob, 1e-7, 1.0 - 1e-7)
    row_sums = np.clip(prob.sum(axis=1, keepdims=True), 1e-12, None)
    prob = prob / row_sums

    pred = prob.argmax(axis=1)
    n_classes = prob.shape[1]

    if n_classes == 2:
        roc = float(roc_auc_score(y_true, prob[:, 1]))
    else:
        roc = float(roc_auc_score(y_true, prob, multi_class="ovr", average="macro"))

    return {
        "roc_auc": roc,
        "accuracy": float(accuracy_score(y_true, pred)),
        "precision": float(precision_score(y_true, pred, average="macro", zero_division=0)),
        "recall": float(recall_score(y_true, pred, average="macro", zero_division=0)),
        "f1": float(f1_score(y_true, pred, average="macro", zero_division=0)),
        "log_loss": float(log_loss(y_true, prob, labels=np.arange(n_classes))),
        "ece": float(expected_calibration_error_toplabel(y_true, prob, n_bins=N_ECE_BINS)),
    }


def print_epoch_forward_metrics(
    epoch: int,
    loss_value: float | None,
    lr: float | None,
    y_true: np.ndarray,
    prob: np.ndarray,
    y_val: np.ndarray | None = None,
    prob_val: np.ndarray | None = None,
):
    m = compute_classification_metrics(y_true, prob)
    loss_text = f"{loss_value:.5f}" if loss_value is not None else "NA"
    lr_text = f"{lr:.6f}" if lr is not None else "NA"

    if y_val is not None and prob_val is not None:
        vm = compute_classification_metrics(y_val, prob_val)
        val_loss = vm["log_loss"]
        print(
            f"Epoch {epoch:03d} | Loss: {loss_text} | "
            f"Accuracy: {m['accuracy']:.5f} | Precision: {m['precision']:.5f} | Recall: {m['recall']:.5f} | "
            f"F1: {m['f1']:.5f} | ROC-AUC: {m['roc_auc']:.5f} | "
            f"Val Loss: {val_loss:.5f} | Val Accuracy: {vm['accuracy']:.5f} | Val Precision: {vm['precision']:.5f} | "
            f"Val Recall: {vm['recall']:.5f} | Val F1: {vm['f1']:.5f} | Val ROC-AUC: {vm['roc_auc']:.5f} | "
            f"Val CE: {vm['log_loss']:.5f} | Val ECE: {vm['ece']:.5f} | LR: {lr_text}"
        )
    else:
        print(
            f"Epoch {epoch:03d} | Loss: {loss_text} | "
            f"Accuracy: {m['accuracy']:.5f} | Precision: {m['precision']:.5f} | Recall: {m['recall']:.5f} | "
            f"F1: {m['f1']:.5f} | ROC-AUC: {m['roc_auc']:.5f} | "
            f"Val Loss: NA | Val Accuracy: NA | Val Precision: NA | Val Recall: NA | Val F1: NA | "
            f"Val ROC-AUC: NA | Val CE: NA | Val ECE: NA | LR: {lr_text}"
        )


def predict_proba_torch_batched(model, X_np: np.ndarray, batch_size: int = 1024) -> np.ndarray:
    model.eval()
    out = []
    with torch.no_grad():
        for i in range(0, len(X_np), batch_size):
            xb = torch.tensor(X_np[i : i + batch_size], dtype=torch.float32, device=DEVICE)
            logits = model(xb)
            out.append(torch.softmax(logits, dim=-1).detach().cpu().numpy())
    return np.vstack(out)


def predict_proba_tabularsetssm_context_batched(
    model,
    X_context: np.ndarray,
    y_context: np.ndarray,
    X_query: np.ndarray,
    query_batch_size: int = 2048,
    max_context_rows: int | None = None,
) -> np.ndarray:
    """Use TabularSetSSM context API with chunked queries for large datasets."""
    model.eval()
    if max_context_rows is not None and len(X_context) > int(max_context_rows):
        idx = np.random.RandomState(seed).choice(len(X_context), size=int(max_context_rows), replace=False)
        X_context = X_context[idx]
        y_context = y_context[idx]
    out = []
    with torch.no_grad():
        for i in range(0, len(X_query), query_batch_size):
            q = X_query[i : i + query_batch_size]
            pr = model.predict_proba_with_context(X_context, y_context, q)
            out.append(np.asarray(pr, dtype=float))
    return np.vstack(out)


def predict_proba_sklearn_batched(model, X_np: np.ndarray, batch_size: int = 8192, **kwargs) -> np.ndarray:
    """Batched ``predict_proba`` with robust output shaping and CUDA-aware XGBoost fallback."""
    xgb_cls_name = "XGBClassifier"
    try:
        from xgboost import XGBClassifier as _XGBC  # noqa: F401

        xgb_cls = type(model).__name__ == xgb_cls_name or isinstance(model, _XGBC)
    except Exception:
        xgb_cls = type(model).__name__ == xgb_cls_name
    dev = None
    if xgb_cls:
        try:
            dev = model.get_params().get("device", None)
        except Exception:
            dev = None
        if dev is None:
            dev = getattr(model, "device", None)
    use_cuda = dev is not None and str(dev).lower().startswith("cuda")

    if use_cuda:
        try:
            import cupy as cp
        except ImportError:
            cp = None

        if cp is not None:
            out = []
            for i in range(0, len(X_np), batch_size):
                xb = cp.asarray(X_np[i : i + batch_size], dtype=cp.float32)
                pr = np.asarray(cp.asnumpy(model.predict_proba(xb, **kwargs)), dtype=float)
                if pr.ndim == 1:
                    pr = pr.reshape(-1, 1)
                out.append(pr)
            return np.concatenate(out, axis=0) if out else np.empty((0, 0), dtype=float)

        # If CuPy is unavailable, avoid sklearn wrapper inplace_predict CPU->GPU mismatch warnings
        # by calling Booster.predict over DMatrix directly.
        if xgb_cls:
            try:
                import xgboost as xgb
            except Exception:
                xgb = None
            if xgb is not None:
                out = []
                booster = model.get_booster()
                iter_range = kwargs.get("iteration_range", None)
                for i in range(0, len(X_np), batch_size):
                    dm = xgb.DMatrix(np.asarray(X_np[i : i + batch_size], dtype=np.float32))
                    if iter_range is not None:
                        pr = booster.predict(dm, iteration_range=iter_range)
                    else:
                        pr = booster.predict(dm)
                    pr = np.asarray(pr, dtype=float)
                    if pr.ndim == 1:
                        pr = pr.reshape(-1, 1)
                    out.append(pr)
                return np.concatenate(out, axis=0) if out else np.empty((0, 0), dtype=float)

    out = []
    for i in range(0, len(X_np), batch_size):
        pr = np.asarray(model.predict_proba(X_np[i : i + batch_size], **kwargs), dtype=float)
        if pr.ndim == 1:
            pr = pr.reshape(-1, 1)
        out.append(pr)
    return np.concatenate(out, axis=0) if out else np.empty((0, 0), dtype=float)


def align_sklearn_proba_to_num_classes(
    prob: np.ndarray,
    classes_ordered: np.ndarray,
    label_universe: np.ndarray,
    expected_n_samples: int | None = None,
) -> np.ndarray:
    """
    Map ``predict_proba`` columns (ordered by ``classes_ordered`` from the fitted estimator)
    to probability columns aligned with ``label_universe`` (sorted unique dataset labels).
    Column ``k`` of the output corresponds to ``label_universe[k]`` (for sklearn metrics).
    """
    prob = np.asarray(prob, dtype=float)
    label_universe = np.asarray(label_universe, dtype=int).ravel()
    K = int(label_universe.size)
    if K == 0:
        raise ValueError("label_universe is empty")
    idx_of = {int(lab): i for i, lab in enumerate(label_universe)}
    classes_ordered = np.asarray(classes_ordered).ravel()

    # Some XGBoost predict paths return 1D for multiclass:
    # - flattened probs (n_samples * K,)
    # - predicted class indices (n_samples,)
    # - single-class score vectors (n_samples,)
    if prob.ndim == 1 and K > 2:
        n_exp = int(expected_n_samples) if expected_n_samples is not None else None
        if n_exp is not None and n_exp > 0:
            if prob.size == n_exp * K:
                prob = prob.reshape(n_exp, K)
            elif classes_ordered.size > 0 and prob.size == n_exp * classes_ordered.size:
                prob = prob.reshape(n_exp, classes_ordered.size)
            elif prob.size == n_exp:
                # Keep as sample-wise vector and convert below.
                pass
            elif prob.size % K == 0:
                prob = prob.reshape(-1, K)
            elif classes_ordered.size > 0 and prob.size % classes_ordered.size == 0:
                prob = prob.reshape(-1, classes_ordered.size)
        else:
            if prob.size % K == 0:
                prob = prob.reshape(-1, K)
            elif classes_ordered.size > 0 and prob.size % classes_ordered.size == 0:
                prob = prob.reshape(-1, classes_ordered.size)

        if prob.ndim == 1:
            # If 1D values are integer class predictions, convert to one-hot probabilities.
            p_int = np.rint(prob).astype(int)
            cls_set = set(int(c) for c in np.asarray(classes_ordered, dtype=int).ravel())
            if prob.size > 0 and np.allclose(prob, p_int) and set(p_int.tolist()).issubset(cls_set):
                out = np.full((prob.size, K), 1.0 / K, dtype=float)
                for i, lab in enumerate(p_int.tolist()):
                    col = idx_of.get(int(lab), None)
                    if col is not None:
                        out[i, :] = 1e-7
                        out[i, col] = 1.0 - (K - 1) * 1e-7
                out = out / np.clip(out.sum(axis=1, keepdims=True), 1e-12, None)
                return np.clip(out, 1e-7, 1.0 - 1e-7)

            # Last-resort: treat this as one class score and distribute remainder.
            anchor_lab = int(classes_ordered[-1]) if classes_ordered.size > 0 else int(label_universe[-1])
            anchor_col = idx_of.get(anchor_lab, K - 1)
            p1 = np.clip(prob.astype(float), 1e-7, 1.0 - 1e-7)
            out = np.full((p1.size, K), 0.0, dtype=float)
            rest = (1.0 - p1) / max(K - 1, 1)
            out[:, :] = rest[:, None]
            out[:, anchor_col] = p1
            out = out / np.clip(out.sum(axis=1, keepdims=True), 1e-12, None)
            return np.clip(out, 1e-7, 1.0 - 1e-7)

    if prob.ndim == 1:
        if K != 2 or classes_ordered.size != 2:
            raise ValueError(
                "1D predict_proba expects binary problems with two known classes; "
                f"got K={K}, len(classes_ordered)={classes_ordered.size}"
            )
        c0, c1 = int(classes_ordered[0]), int(classes_ordered[1])
        out = np.full((prob.shape[0], K), 1.0 / K, dtype=float)
        out[:, idx_of[c0]] = 1.0 - prob
        out[:, idx_of[c1]] = prob
        out = out / np.clip(out.sum(axis=1, keepdims=True), 1e-12, None)
        return np.clip(out, 1e-7, 1.0 - 1e-7)

    # If upstream reshaped flattened 1D output into a single-column matrix, recover it.
    if prob.ndim == 2 and prob.shape[1] == 1 and K > 2 and prob.shape[0] % K == 0:
        prob = prob.reshape(-1, K)

    if prob.ndim == 2 and prob.shape[1] == 1 and K == 2 and classes_ordered.size == 2:
        p1 = prob[:, 0]
        c0, c1 = int(classes_ordered[0]), int(classes_ordered[1])
        out = np.full((prob.shape[0], K), 1.0 / K, dtype=float)
        out[:, idx_of[c0]] = 1.0 - p1
        out[:, idx_of[c1]] = p1
        out = out / np.clip(out.sum(axis=1, keepdims=True), 1e-12, None)
        return np.clip(out, 1e-7, 1.0 - 1e-7)

    if prob.ndim != 2:
        raise ValueError(f"predict_proba must be 1D or 2D after transforms, got shape {prob.shape}")

    out = np.full((prob.shape[0], K), 1.0 / max(K, 1), dtype=float)
    n_cols = min(prob.shape[1], classes_ordered.size)
    for j in range(n_cols):
        lab = int(classes_ordered[j])
        if lab not in idx_of:
            continue
        out[:, idx_of[lab]] = prob[:, j]
    out = out / np.clip(out.sum(axis=1, keepdims=True), 1e-12, None)
    return np.clip(out, 1e-7, 1.0 - 1e-7)


def extract_prob_from_ptt_prediction(pred_df: pd.DataFrame, num_classes: int) -> np.ndarray:
    cols = list(pred_df.columns)

    prob_cols = [c for c in cols if "probability" in str(c).lower()]
    if prob_cols:
        arr = pred_df[prob_cols].to_numpy(dtype=np.float32)
        if arr.shape[1] == num_classes:
            return arr
        if arr.shape[1] > num_classes:
            return arr[:, :num_classes]

    numeric_cols = pred_df.select_dtypes(include=[np.number]).columns.tolist()
    if len(numeric_cols) >= num_classes:
        arr = pred_df[numeric_cols[:num_classes]].to_numpy(dtype=np.float32)
        row_sums = np.clip(arr.sum(axis=1, keepdims=True), 1e-12, None)
        return arr / row_sums

    pred_col = next((c for c in cols if str(c).lower() in {"prediction", "pred", "preds"}), None)
    if pred_col is not None:
        pred_vals = pred_df[pred_col].to_numpy()
        try:
            pred_idx = pred_vals.astype(int)
        except ValueError:
            pred_idx = pd.Categorical(pred_vals).codes
        pred_idx = np.clip(pred_idx, 0, num_classes - 1)
        out = np.zeros((len(pred_df), num_classes), dtype=np.float32)
        out[np.arange(len(pred_df)), pred_idx] = 1.0
        return out

    raise ValueError(f"Unable to parse pytorch-tabular predictions; columns={cols}")


def predict_proba_ptt_batched(model, X_np: np.ndarray, feature_cols: list[str], num_classes: int, batch_size: int = 8192) -> np.ndarray:
    out = []
    for i in range(0, len(X_np), batch_size):
        batch_df = pd.DataFrame(X_np[i : i + batch_size], columns=feature_cols)
        pred_df = model.predict(batch_df)
        out.append(extract_prob_from_ptt_prediction(pred_df, num_classes))
    return np.vstack(out)


def fit_preprocessor(X_df: pd.DataFrame) -> dict:
    X_fit = X_df.copy()
    num_cols = [c for c in X_fit.columns if pd.api.types.is_numeric_dtype(X_fit[c])]
    cat_cols = [c for c in X_fit.columns if c not in num_cols]

    num_fill = {c: float(X_fit[c].median()) if not X_fit[c].dropna().empty else 0.0 for c in num_cols}

    for c in num_cols:
        X_fit[c] = X_fit[c].fillna(num_fill[c])
    for c in cat_cols:
        X_fit[c] = X_fit[c].astype(str).fillna("__MISSING__")

    X_enc = pd.get_dummies(X_fit, columns=cat_cols, drop_first=False)
    columns = list(X_enc.columns)

    return {
        "num_cols": num_cols,
        "cat_cols": cat_cols,
        "num_fill": num_fill,
        "columns": columns,
    }


def transform_with_preprocessor(X_df: pd.DataFrame, prep: dict, do_scale: bool = False, scaler: Any = None, fit_scaler: bool = False):
    X_t = X_df.copy()

    for c in prep["num_cols"]:
        if c not in X_t.columns:
            X_t[c] = prep["num_fill"][c]
        X_t[c] = X_t[c].fillna(prep["num_fill"][c])

    for c in prep["cat_cols"]:
        if c not in X_t.columns:
            X_t[c] = "__MISSING__"
        X_t[c] = X_t[c].astype(str).fillna("__MISSING__")

    X_enc = pd.get_dummies(X_t, columns=prep["cat_cols"], drop_first=False)
    X_enc = X_enc.reindex(columns=prep["columns"], fill_value=0.0)
    X_np = X_enc.to_numpy(dtype=np.float32)

    if do_scale:
        if fit_scaler:
            scaler = StandardScaler()
            X_np = scaler.fit_transform(X_np).astype(np.float32)
        else:
            X_np = scaler.transform(X_np).astype(np.float32)

    return X_np, scaler


def predict_proba_torch_batched(model, X_np: np.ndarray, batch_size: int = 1024) -> np.ndarray:
    model.eval()
    out = []
    with torch.no_grad():
        for i in range(0, len(X_np), batch_size):
            xb = torch.tensor(X_np[i : i + batch_size], dtype=torch.float32, device=DEVICE)
            logits = model(xb)
            prob = torch.softmax(logits, dim=-1)[:, 1].detach().cpu().numpy()
            out.append(prob)
    return np.concatenate(out, axis=0)


def predict_proba_sklearn_batched(model, X_np: np.ndarray, batch_size: int = 8192, **kwargs) -> np.ndarray:
    out = []
    for i in range(0, len(X_np), batch_size):
        p = model.predict_proba(X_np[i : i + batch_size], **kwargs)
        out.append(p[:, 1])
    return np.concatenate(out, axis=0)


def extract_binary_prob_from_ptt_prediction(pred_df: pd.DataFrame) -> np.ndarray:
    cols = list(pred_df.columns)
    prob_cols = [c for c in cols if "probability" in str(c).lower()]
    if prob_cols:
        arr = pred_df[prob_cols].to_numpy(dtype=np.float32)
        if arr.shape[1] == 1:
            return arr[:, 0]
        if arr.shape[1] >= 2:
            return arr[:, 1]

    numeric_cols = pred_df.select_dtypes(include=[np.number]).columns.tolist()
    if len(numeric_cols) >= 2:
        arr = pred_df[numeric_cols].to_numpy(dtype=np.float32)
        row_sums = np.clip(arr.sum(axis=1, keepdims=True), 1e-12, None)
        arr = arr / row_sums
        return arr[:, 1]

    pred_col = next((c for c in cols if str(c).lower() in {"prediction", "pred", "preds"}), None)
    if pred_col is not None:
        pred_vals = pred_df[pred_col].to_numpy()
        try:
            y_hat = pred_vals.astype(int)
        except ValueError:
            y_hat = pd.Categorical(pred_vals).codes
        return y_hat.astype(np.float32)

    raise ValueError(f"Unable to parse pytorch-tabular predictions; columns={cols}")


def checkpoint_path(model_name: str, dataset_key: str, fold_idx: int) -> Path:
    model_dir = CHECKPOINT_ROOT / model_name
    model_dir.mkdir(parents=True, exist_ok=True)
    safe_ds = dataset_key.replace("/", "_").replace(" ", "_")
    return model_dir / f"{safe_ds}_fold{fold_idx}_best_model_{timestamp_now()}.pth"


def get_benchmark_registry(registry: list[dict]) -> list[dict]:
    """Return all registry entries (any number of classes), order preserved."""
    return list(registry)


def build_shared_folds(y: np.ndarray) -> list[tuple[np.ndarray, np.ndarray]]:
    skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=seed)
    return list(skf.split(np.zeros(len(y)), y))


def run_single_fold(
    model_name: str,
    dataset_key: str,
    fold_idx: int,
    X_df: pd.DataFrame,
    y: np.ndarray,
    train_idx: np.ndarray,
    test_idx: np.ndarray,
) -> tuple[dict, float]:
    t0 = time.perf_counter()

    X_train_fold_df = X_df.iloc[train_idx].reset_index(drop=True)
    y_train_fold = y[train_idx]
    X_test_fold_df = X_df.iloc[test_idx].reset_index(drop=True)
    y_test_fold = y[test_idx]

    tr_sub_idx, val_sub_idx = train_test_split(
        np.arange(len(y_train_fold)),
        test_size=INNER_VAL_RATIO,
        random_state=seed + fold_idx,
        stratify=y_train_fold,
    )

    X_tr_df = X_train_fold_df.iloc[tr_sub_idx].reset_index(drop=True)
    y_tr = y_train_fold[tr_sub_idx]
    X_val_df = X_train_fold_df.iloc[val_sub_idx].reset_index(drop=True)
    y_val = y_train_fold[val_sub_idx]

    prep = fit_preprocessor(X_tr_df)

    do_scale = model_name in {"tabularsetssm"}
    X_tr, scaler = transform_with_preprocessor(X_tr_df, prep, do_scale=do_scale, scaler=None, fit_scaler=True)
    X_val, _ = transform_with_preprocessor(X_val_df, prep, do_scale=do_scale, scaler=scaler, fit_scaler=False)
    X_test, _ = transform_with_preprocessor(X_test_fold_df, prep, do_scale=do_scale, scaler=scaler, fit_scaler=False)

    label_universe = np.sort(np.unique(y)).astype(int)
    num_classes = int(label_universe.size)
    best_val_auc = -np.inf
    best_epoch = 1
    test_prob = None
    payload = {}

    if model_name == "tabularsetssm":
        p = MODEL_PARAMS["tabularsetssm"]
        # Cache one LR per dataset: first fold can auto-find, later folds reuse it.
        if "_TABULARSETSSM_DATASET_LR_CACHE" not in globals():
            _TABULARSETSSM_DATASET_LR_CACHE = {}
        cached_lr = _TABULARSETSSM_DATASET_LR_CACHE.get(dataset_key, None)

        requested_lr = p.get("learning_rate", None)
        use_lr = cached_lr if cached_lr is not None else requested_lr
        # Explicit learning_rate always takes precedence over auto_lr.
        auto_lr_flag = (requested_lr is None) and p.get("auto_lr", True) and (cached_lr is None)

        cfg = TabularSetSSMConfig(
            num_features=X_tr.shape[1],
            hidden_dim=p["hidden_dim"],
            pe_dim=p["pe_dim"],
            pe_num_filters=p["pe_num_filters"],
            rank_dim=p["rank_dim"],
            num_blocks=p["num_blocks"],
            use_selective=p.get("use_selective", True),
            dropout=p["dropout"],
            graph_method=p.get("graph_method", "correlation"),
            activation=p.get("activation", "gelu"),
            learning_rate=use_lr,
            default_learning_rate=p.get("default_learning_rate", 1e-3),
            weight_decay=p["weight_decay"],
            batch_size=p["batch_size"],
            max_epochs=p["max_epochs"],
            patience=p.get("patience", EARLY_STOPPING_PATIENCE),
            val_ratio=p.get("val_ratio"),
            val_stratify=p.get("val_stratify", False),
            seed=seed,
            device=DEVICE,
            save_best_checkpoint=True,
            lr_verbose=p.get("lr_verbose", True),
            epoch_metrics_verbose=p.get("epoch_metrics_verbose", True),
            scheduler_patience=p.get("scheduler_patience", 5),
            scheduler_factor=p.get("scheduler_factor", 0.5),
            scheduler_min_lr=p.get("scheduler_min_lr", 1e-6),
            scheduler_monitor=p.get("scheduler_monitor", "train_loss"),
            lr_finder_end_lr=p.get("lr_finder_end_lr", 10.0),
            lr_finder_num_iter=p.get("lr_finder_num_iter", 100),
            auto_lr=auto_lr_flag,
            soft_f1_lambda=p.get("soft_f1_lambda", 0.5),
            temperature_lr=p.get("temperature_lr", 0.05),
            temperature_tune_steps=p.get("temperature_tune_steps", 1),
            max_experts=p.get("max_experts", 32),
            tau_start=p.get("tau_start", 1.0),
            tau_end=p.get("tau_end", 0.01),
            anneal_epochs=p.get("anneal_epochs", 20),
            lambda_balance=p.get("lambda_balance", 0.01),
            lambda_sparse=p.get("lambda_sparse", 0.001),
            expert_usage_threshold=p.get("expert_usage_threshold", 0.01),
            hard_routing_after_anneal=p.get("hard_routing_after_anneal", True),
            router_hidden_dim=p.get("router_hidden_dim", 64),
            router_top_k=p.get("router_top_k", 2),
            router_top_k_train_mode=p.get("router_top_k_train_mode", "soft_topk"),
            expert_use_glu_ffn=p.get("expert_use_glu_ffn", True),
            expert_ffn_mult=p.get("expert_ffn_mult", 2.0),
            expert_use_feature_cross=p.get("expert_use_feature_cross", False),
            expert_cross_rank=p.get("expert_cross_rank", 16),
        )
        model = TabularSetSSM(config=cfg, task_type="classification", num_classes=num_classes).to(DEVICE)
        model.task_type_ = "classification"

        # Same 20% stratified fold val as other models: val ROC-AUC early stop + per-epoch T tuning (see TabularSetSSM.fit).
        model.fit(X_tr, y_tr, x_val=X_val, y_val=y_val)

        start_lr = float(model.current_learning_rate or cfg.learning_rate)
        if cached_lr is None:
            _TABULARSETSSM_DATASET_LR_CACHE[dataset_key] = start_lr
            print(f"[TabularSetSSM] Cached dataset LR for {dataset_key}: {start_lr:.8f}")
        else:
            print(f"[TabularSetSSM] Reusing cached LR for {dataset_key}: {start_lr:.8f}")

        use_context_inference = bool(p.get("use_context_inference", False))
        if use_context_inference:
            val_prob = predict_proba_tabularsetssm_context_batched(
                model,
                X_tr,
                y_tr,
                X_val,
                query_batch_size=p.get("context_query_batch_size", 2048),
                max_context_rows=p.get("max_context_rows", 50000),
            )
        else:
            val_prob = predict_proba_torch_batched(model, X_val, batch_size=1024)
        best_val_auc = compute_classification_metrics(y_val, val_prob)["roc_auc"]

        # Primary source: trainer sets this from best val ROC-AUC epoch under early stopping.
        best_epoch = int(getattr(model, "best_training_epoch_", 0) or 0)
        if best_epoch <= 0 and getattr(model, "best_checkpoint_path", None):
            try:
                ckpt_payload = torch.load(model.best_checkpoint_path, map_location=DEVICE)
                best_epoch = int(ckpt_payload.get("best_epoch", 0) or 0)
            except Exception:
                best_epoch = 0
        if best_epoch <= 0:
            best_epoch = int(getattr(cfg, "max_epochs", 1))

        print(f"[TabularSetSSM] Fit complete | best_epoch={best_epoch} | start_lr={start_lr:.8f}")

        if use_context_inference:
            test_prob = predict_proba_tabularsetssm_context_batched(
                model,
                X_tr,
                y_tr,
                X_test,
                query_batch_size=p.get("context_query_batch_size", 2048),
                max_context_rows=p.get("max_context_rows", 50000),
            )
        else:
            test_prob = predict_proba_torch_batched(model, X_test, batch_size=1024)
        payload = {
            "config": asdict(cfg),
            "model_state_dict": model.state_dict(),
            "best_epoch": best_epoch,
            "best_val_auc": float(best_val_auc),
            "best_checkpoint_path": getattr(model, "best_checkpoint_path", None),
            "start_learning_rate": float(model.current_learning_rate or cfg.learning_rate),
        }

    elif model_name == "catboost":
        p = MODEL_PARAMS["catboost"]
        loss_fn = "Logloss" if num_classes == 2 else "MultiClass"
        eval_metric = "AUC"

        _cb_core = dict(
            iterations=p["iterations"],
            learning_rate=p["learning_rate"],
            depth=p["depth"],
            task_type=p["task_type"],
            loss_function=loss_fn,
            eval_metric=eval_metric,
            random_seed=seed,
            verbose=False,
            od_type="Iter",
            od_wait=p.get("od_wait", EARLY_STOPPING_PATIENCE),
        )
        model = CatBoostClassifier(**{**_p_extras(p, *_cb_core.keys()), **_cb_core})
        model.fit(X_tr, y_tr, eval_set=(X_val, y_val), use_best_model=True, verbose=False)

        val_eval_start = 1
        no_improve = 0
        trained_trees = int(getattr(model, "tree_count_", p["iterations"]))

        staged_val_iter = model.staged_predict_proba(X_val)
        for epoch, tr_prob in enumerate(model.staged_predict_proba(X_tr), start=1):
            tr_prob = np.asarray(tr_prob)
            tr_loss = float(log_loss(y_tr, np.clip(tr_prob, 1e-7, 1.0 - 1e-7), labels=label_universe))
            va_prob = np.asarray(next(staged_val_iter)) if epoch <= trained_trees else np.asarray(model.predict_proba(X_val))
            print_epoch_forward_metrics(epoch, tr_loss, p["learning_rate"], y_tr, tr_prob, y_val, va_prob)

            if epoch >= val_eval_start:
                val_auc = compute_classification_metrics(y_val, va_prob)["roc_auc"]
                if val_auc > best_val_auc:
                    best_val_auc = val_auc
                    best_epoch = epoch
                    no_improve = 0
                else:
                    no_improve += 1
                if no_improve >= EARLY_STOPPING_PATIENCE:
                    break

        if best_epoch <= 0:
            best_epoch = trained_trees
        test_prob = predict_proba_sklearn_batched(model, X_test, ntree_end=best_epoch)
        payload = {"model": model, "best_epoch": best_epoch, "best_val_auc": float(best_val_auc)}

    elif model_name == "xgboost":
        p = MODEL_PARAMS["xgboost"]
        _xgb_device = str(p.get("device", "cuda"))
        if not _xgb_device.lower().startswith("cuda"):
            raise ValueError(f"XGBoost is configured for '{_xgb_device}', but this benchmark requires GPU (cuda).")
        try:
            import torch
            if not torch.cuda.is_available():
                raise RuntimeError(
                    "XGBoost GPU run requested (device='cuda') but no CUDA device is available."
                )
        except Exception as e:
            if isinstance(e, RuntimeError):
                raise

        xgb_kwargs = {
            "eval_metric": "auc",
            "n_estimators": p["n_estimators"],
            "learning_rate": p["learning_rate"],
            "max_depth": p["max_depth"],
            "subsample": p["subsample"],
            "colsample_bytree": p["colsample_bytree"],
            "random_state": seed,
            "tree_method": "hist",
            "device": _xgb_device,
            "early_stopping_rounds": EARLY_STOPPING_PATIENCE,
        }
        if num_classes == 2:
            xgb_kwargs["objective"] = "binary:logistic"
        else:
            xgb_kwargs["objective"] = "multi:softprob"
            xgb_kwargs["num_class"] = num_classes

        model = XGBClassifier(**{**_p_extras(p, *xgb_kwargs.keys()), **xgb_kwargs})
        # For tree boosting, one boosting round is the model's epoch.
        # eval_metric='auc' evaluates validation ROC-AUC each round and early stopping tracks this metric.
        model.fit(X_tr, y_tr, eval_set=[(X_val, y_val)], verbose=False)

        cls_order = np.asarray(model.classes_, dtype=int)

        try:
            trained_rounds = int(model.get_booster().num_boosted_rounds())
        except Exception:
            trained_rounds = int(getattr(model, "best_iteration", 0)) + EARLY_STOPPING_PATIENCE + 1
        if trained_rounds <= 0:
            trained_rounds = p["n_estimators"]

        best_epoch = 1
        best_val_auc = -np.inf
        no_improve = 0

        for epoch in range(1, trained_rounds + 1):
            tr_prob = align_sklearn_proba_to_num_classes(
                predict_proba_sklearn_batched(model, X_tr, iteration_range=(0, epoch)),
                cls_order,
                label_universe,
                expected_n_samples=len(y_tr),
            )
            va_prob = align_sklearn_proba_to_num_classes(
                predict_proba_sklearn_batched(model, X_val, iteration_range=(0, epoch)),
                cls_order,
                label_universe,
                expected_n_samples=len(y_val),
            )
            tr_loss = float(log_loss(y_tr, tr_prob, labels=label_universe))
            print_epoch_forward_metrics(epoch, tr_loss, p["learning_rate"], y_tr, tr_prob, y_val, va_prob)

            val_auc = compute_classification_metrics(y_val, va_prob)["roc_auc"]
            if val_auc > best_val_auc:
                best_val_auc = float(val_auc)
                best_epoch = epoch
                no_improve = 0
            else:
                no_improve += 1
            if no_improve >= EARLY_STOPPING_PATIENCE:
                break

        val_prob = align_sklearn_proba_to_num_classes(
            predict_proba_sklearn_batched(model, X_val, iteration_range=(0, best_epoch)),
            cls_order,
            label_universe,
            expected_n_samples=len(y_val),
        )
        best_val_auc = compute_classification_metrics(y_val, val_prob)["roc_auc"]
        test_prob = align_sklearn_proba_to_num_classes(
            predict_proba_sklearn_batched(model, X_test, iteration_range=(0, best_epoch)),
            cls_order,
            label_universe,
            expected_n_samples=len(y_test_fold),
        )
        payload = {"model": model, "best_epoch": best_epoch, "best_val_auc": float(best_val_auc)}

    elif model_name == "lightgbm":
        import lightgbm as lgb
        p = MODEL_PARAMS["lightgbm"]

        lgb_kwargs = {
            "n_estimators": p["n_estimators"],
            "learning_rate": p["learning_rate"],
            "random_state": seed,
        }
        if num_classes == 2:
            lgb_kwargs["objective"] = "binary"
        else:
            lgb_kwargs["objective"] = "multiclass"
            lgb_kwargs["num_class"] = num_classes

        model = LGBMClassifier(**{**_p_extras(p, *lgb_kwargs.keys()), **lgb_kwargs})
        model.fit(
            X_tr,
            y_tr,
            eval_set=[(X_val, y_val)],
            eval_metric="auc_mu" if num_classes > 2 else "auc",
            callbacks=[lgb.early_stopping(stopping_rounds=EARLY_STOPPING_PATIENCE, first_metric_only=True, verbose=False)],
        )

        cls_order = np.asarray(model.classes_, dtype=int)

        best_epoch = int(getattr(model, "best_iteration_", 0) or 0)
        if best_epoch <= 0:
            best_epoch = int(getattr(model, "n_estimators_", 1))

        for epoch in range(1, best_epoch + 1):
            tr_prob = align_sklearn_proba_to_num_classes(
                predict_proba_sklearn_batched(model, X_tr, num_iteration=epoch),
                cls_order,
                label_universe,
            )
            va_prob = align_sklearn_proba_to_num_classes(
                predict_proba_sklearn_batched(model, X_val, num_iteration=epoch),
                cls_order,
                label_universe,
            )
            tr_loss = float(log_loss(y_tr, tr_prob, labels=label_universe))
            print_epoch_forward_metrics(epoch, tr_loss, p["learning_rate"], y_tr, tr_prob, y_val, va_prob)

        val_prob = align_sklearn_proba_to_num_classes(
            predict_proba_sklearn_batched(model, X_val, num_iteration=best_epoch),
            cls_order,
            label_universe,
        )
        best_val_auc = compute_classification_metrics(y_val, val_prob)["roc_auc"]
        test_prob = align_sklearn_proba_to_num_classes(
            predict_proba_sklearn_batched(model, X_test, num_iteration=best_epoch),
            cls_order,
            label_universe,
        )
        payload = {"model": model, "best_epoch": best_epoch, "best_val_auc": float(best_val_auc)}


    elif model_name == "tabnet":
        p = MODEL_PARAMS["tabnet"]
        _tn_core = dict(optimizer_params={"lr": p["learning_rate"]}, seed=seed, verbose=0)
        model = TabNetClassifier(**{**_p_extras(p, *_tn_core.keys(), "learning_rate", "max_epochs", "batch_size", "virtual_batch_size"), **_tn_core})

        max_epochs = p["max_epochs"]
        val_eval_start = validation_eval_start_epoch(max_epochs)
        no_improve = 0
        best_model_obj = None

        for epoch in range(1, max_epochs + 1):
            model.fit(
                X_train=X_tr,
                y_train=y_tr,
                eval_set=[(X_val, y_val)],
                eval_metric=["auc" if num_classes == 2 else "logloss"],
                max_epochs=1,
                patience=0,
                batch_size=p["batch_size"],
                virtual_batch_size=p["virtual_batch_size"],
                num_workers=0,
                drop_last=False,
                warm_start=(epoch > 1),
            )

            tr_prob = predict_proba_sklearn_batched(model, X_tr)
            val_prob = predict_proba_sklearn_batched(model, X_val)
            tr_loss = float(log_loss(y_tr, np.clip(tr_prob, 1e-7, 1.0 - 1e-7), labels=label_universe))
            current_lr = None
            if hasattr(model, "_optimizer") and model._optimizer is not None:
                current_lr = float(model._optimizer.param_groups[0]["lr"])
            print_epoch_forward_metrics(
                epoch,
                tr_loss,
                current_lr if current_lr is not None else p["learning_rate"],
                y_tr,
                tr_prob,
                y_val,
                val_prob,
            )

            if epoch >= val_eval_start:
                val_auc = compute_classification_metrics(y_val, val_prob)["roc_auc"]
                if val_auc > best_val_auc:
                    best_val_auc = val_auc
                    best_epoch = epoch
                    best_model_obj = copy.deepcopy(model)
                    no_improve = 0
                else:
                    no_improve += 1
                if no_improve >= EARLY_STOPPING_PATIENCE:
                    break

        if best_model_obj is None:
            best_model_obj = model
            if best_val_auc < 0:
                val_prob = predict_proba_sklearn_batched(best_model_obj, X_val)
                best_val_auc = compute_classification_metrics(y_val, val_prob)["roc_auc"]
                best_epoch = 1

        test_prob = predict_proba_sklearn_batched(best_model_obj, X_test)
        payload = {"model": best_model_obj, "best_epoch": best_epoch, "best_val_auc": float(best_val_auc)}


    elif model_name == "ebm":
        p = MODEL_PARAMS["ebm"]
        _ebm_core = dict(random_state=seed, learning_rate=p["learning_rate"], max_rounds=p["max_rounds"])
        model = ExplainableBoostingClassifier(**{**_p_extras(p, *_ebm_core.keys()), **_ebm_core})
        model.fit(X_tr, y_tr)

        if hasattr(model, "staged_predict_proba"):
            tr_gen = model.staged_predict_proba(X_tr)
            val_gen = model.staged_predict_proba(X_val)
            test_gen = model.staged_predict_proba(X_test)
            val_eval_start = validation_eval_start_epoch(p["max_rounds"])
            chosen_test = None
            no_improve = 0
            for epoch, (tr_prob_mat, va_prob_mat, te_prob_mat) in enumerate(zip(tr_gen, val_gen, test_gen), start=1):
                tr_prob = np.asarray(tr_prob_mat, dtype=float)
                val_prob_epoch = np.asarray(va_prob_mat, dtype=float)
                tr_loss = float(log_loss(y_tr, np.clip(tr_prob, 1e-7, 1.0 - 1e-7), labels=label_universe))
                print_epoch_forward_metrics(epoch, tr_loss, p["learning_rate"], y_tr, tr_prob, y_val, val_prob_epoch)

                if epoch < val_eval_start:
                    continue
                val_prob = np.asarray(va_prob_mat, dtype=float)
                val_auc = compute_classification_metrics(y_val, val_prob)["roc_auc"]
                if val_auc > best_val_auc:
                    best_val_auc = val_auc
                    best_epoch = epoch
                    chosen_test = np.asarray(te_prob_mat, dtype=float)
                    no_improve = 0
                else:
                    no_improve += 1
                if no_improve >= EARLY_STOPPING_PATIENCE:
                    break
            test_prob = chosen_test if chosen_test is not None else np.asarray(model.predict_proba(X_test), dtype=float)
        else:
            tr_prob = np.asarray(model.predict_proba(X_tr), dtype=float)
            tr_loss = float(log_loss(y_tr, np.clip(tr_prob, 1e-7, 1.0 - 1e-7), labels=label_universe))
            val_prob = np.asarray(model.predict_proba(X_val), dtype=float)
            print_epoch_forward_metrics(1, tr_loss, p["learning_rate"], y_tr, tr_prob, y_val, val_prob)
            best_val_auc = compute_classification_metrics(y_val, val_prob)["roc_auc"]
            best_epoch = 1
            test_prob = np.asarray(model.predict_proba(X_test), dtype=float)

        payload = {"model": model, "best_epoch": best_epoch, "best_val_auc": float(best_val_auc)}

    else:
        raise ValueError(f"Unsupported model: {model_name}")

    test_prob = np.asarray(test_prob, dtype=float)
    fold_metrics = compute_classification_metrics(y_test_fold, test_prob)

    elapsed = float(time.perf_counter() - t0)
    ckpt = checkpoint_path(model_name, dataset_key, fold_idx)
    try:
        if model_name == "tabularsetssm":
            torch.save(payload, ckpt)
        else:
            joblib.dump(payload, ckpt)
    except Exception:
        pass

    return fold_metrics, elapsed


def run_model_benchmark(model_name: str, datasets: list[dict]) -> dict:
    if not AVAILABLE_MODELS.get(model_name, False):
        print(f"Skipping {model_name}: dependency not available.")
        return {}

    all_results = {}

    for meta in datasets:
        dataset_key = format_dataset_key(meta)
        df = pd.read_csv(meta["csv_path"])
        target_col = meta["target_col"]
        df = df.dropna(subset=[target_col]).reset_index(drop=True)

        X_df = df.drop(columns=[target_col])
        y_raw = df[target_col].astype(str).to_numpy()
        y = LabelEncoder().fit_transform(y_raw).astype(int)

        # Deterministic dataset shuffle shared across models.
        # Shuffle indices only (no full DataFrame copy) for memory safety.
        dataset_seed = seed + int(np.frombuffer(dataset_key.encode("utf-8"), dtype=np.uint8).sum())
        rng = np.random.default_rng(dataset_seed)
        shuffled_order = rng.permutation(len(y))

        y_shuffled = y[shuffled_order]
        shuffled_folds = build_shared_folds(y_shuffled)
        folds = [(shuffled_order[tr_idx], shuffled_order[te_idx]) for tr_idx, te_idx in shuffled_folds]

        fold_metrics_list = []
        fold_times = []

        print(f"\n[{model_name}] Dataset: {dataset_key} | rows={len(df)} | classes={len(np.unique(y))}")
        for fold_i, (tr_idx, te_idx) in enumerate(folds, start=1):
            metrics, elapsed = run_single_fold(
                model_name=model_name,
                dataset_key=dataset_key,
                fold_idx=fold_i,
                X_df=X_df,
                y=y,
                train_idx=tr_idx,
                test_idx=te_idx,
            )
            fold_metrics_list.append(metrics)
            fold_times.append(elapsed)
            print(
                f"  Fold {fold_i:02d} Test | "
                f"Accuracy: {metrics['accuracy']:.5f} | Precision: {metrics['precision']:.5f} | "
                f"Recall: {metrics['recall']:.5f} | F1: {metrics['f1']:.5f} | ROC-AUC: {metrics['roc_auc']:.5f} | "
                f"CE: {metrics['log_loss']:.5f} | ECE: {metrics['ece']:.5f} | Time: {elapsed:.2f}s"
            )

        fold_df = pd.DataFrame(fold_metrics_list)
        dataset_mean = fold_df.mean().to_dict()
        dataset_std = fold_df.std(ddof=0).to_dict()

        all_results[dataset_key] = {
            "fold_metrics": fold_metrics_list,
            "dataset_mean": dataset_mean,
            "dataset_std": dataset_std,
            "mean_time_s": float(np.mean(fold_times)),
            "std_time_s": float(np.std(fold_times)),
            "n_rows": int(len(df)),
        }

        print(
            f"  Dataset mean -> ROC={dataset_mean['roc_auc']:.4f}, Acc={dataset_mean['accuracy']:.4f}, "
            f"F1={dataset_mean['f1']:.4f}, CE={dataset_mean['log_loss']:.4f}, ECE={dataset_mean['ece']:.4f}, "
            f"Time={np.mean(fold_times):.2f}s"
        )

    return all_results

benchmark_results = {}
```

## Final Aggregation and Comparison Table

```python
# Build AMLB-style summary table

# Keep only models with non-empty results
valid_models = [m for m in MODEL_ORDER if m in benchmark_results and len(benchmark_results[m]) > 0]
if not valid_models:
    raise ValueError("No model results found. Run at least one model section first.")

# Common datasets across evaluated models
common_datasets = set.intersection(*[set(benchmark_results[m].keys()) for m in valid_models])
common_datasets = sorted(common_datasets)
if not common_datasets:
    raise ValueError("No common datasets across evaluated models.")

print(f"Models in summary: {valid_models}")
print(f"Datasets in summary: {len(common_datasets)}")

# Dataset-level matrix: score[dataset][model][metric] using mean over folds
dataset_metric = {
    ds: {
        m: benchmark_results[m][ds]["dataset_mean"]
        for m in valid_models
    }
    for ds in common_datasets
}

# Mean/Std across datasets for each model+metric
mean_stats = {m: {} for m in valid_models}
std_stats = {m: {} for m in valid_models}
for m in valid_models:
    for metric in METRIC_COLS:
        vals = [dataset_metric[ds][m][metric] for ds in common_datasets]
        mean_stats[m][metric] = float(np.mean(vals))
        std_stats[m][metric] = float(np.std(vals, ddof=0))

# Mean normalized across datasets
norm_stats = {m: {} for m in valid_models}
norm_std_stats = {m: {} for m in valid_models}
for metric in METRIC_COLS:
    per_model_norm_vals = {m: [] for m in valid_models}

    for ds in common_datasets:
        scores = np.array([dataset_metric[ds][m][metric] for m in valid_models], dtype=float)
        s_min, s_max = float(scores.min()), float(scores.max())
        denom = s_max - s_min

        if denom < 1e-12:
            norms = np.full_like(scores, 0.5, dtype=float)
        else:
            if HIGHER_IS_BETTER[metric]:
                norms = (scores - s_min) / denom
            else:
                norms = (s_max - scores) / denom

        for i, m in enumerate(valid_models):
            per_model_norm_vals[m].append(float(norms[i]))

    for m in valid_models:
        vals = np.array(per_model_norm_vals[m], dtype=float)
        norm_stats[m][metric] = float(vals.mean())
        norm_std_stats[m][metric] = float(vals.std(ddof=0))

# Wins counts
wins = {m: {metric: 0 for metric in METRIC_COLS} for m in valid_models}
for ds in common_datasets:
    for metric in METRIC_COLS:
        vals = np.array([dataset_metric[ds][m][metric] for m in valid_models], dtype=float)
        if HIGHER_IS_BETTER[metric]:
            best_val = vals.max()
            winner_ids = np.where(np.isclose(vals, best_val))[0]
        else:
            best_val = vals.min()
            winner_ids = np.where(np.isclose(vals, best_val))[0]

        for idx in winner_ids:
            wins[valid_models[idx]][metric] += 1

# Mean time across datasets (dataset-time means)
time_mean = {}
time_std = {}
for m in valid_models:
    tvals = [benchmark_results[m][ds]["mean_time_s"] for ds in common_datasets]
    time_mean[m] = float(np.mean(tvals))
    time_std[m] = float(np.std(tvals, ddof=0))

# Build display table with grouped columns
def pm(mean_v: float, std_v: float, digits: int = 3) -> str:
    return f"{mean_v:.{digits}f} ±{std_v:.{digits}f}"

col_tuples = [
    ("", "Model"),
    ("Mean Normalized", "ROC (↑)"),
    ("Mean Normalized", "Acc. (↑)"),
    ("Mean Normalized", "F1 (↑)"),
    ("Mean Normalized", "CE (↓)"),
    ("Mean Normalized", "ECE (↓)"),
    ("Mean", "ROC (↑)"),
    ("Mean", "Acc. (↑)"),
    ("Mean", "F1 (↑)"),
    ("Mean", "CE (↓)"),
    ("Mean", "ECE (↓)"),
    ("Wins", "ROC (↑)"),
    ("Wins", "Acc. (↑)"),
    ("Wins", "F1 (↑)"),
    ("Wins", "CE (↓)"),
    ("Wins", "ECE (↓)"),
    ("", "Mean Time (s)"),
]

rows = []
for m in valid_models:
    rows.append([
        m,
        pm(norm_stats[m]["roc_auc"], norm_std_stats[m]["roc_auc"]),
        pm(norm_stats[m]["accuracy"], norm_std_stats[m]["accuracy"]),
        pm(norm_stats[m]["f1"], norm_std_stats[m]["f1"]),
        pm(norm_stats[m]["log_loss"], norm_std_stats[m]["log_loss"]),
        pm(norm_stats[m]["ece"], norm_std_stats[m]["ece"]),
        pm(mean_stats[m]["roc_auc"], std_stats[m]["roc_auc"]),
        pm(mean_stats[m]["accuracy"], std_stats[m]["accuracy"]),
        pm(mean_stats[m]["f1"], std_stats[m]["f1"]),
        pm(mean_stats[m]["log_loss"], std_stats[m]["log_loss"]),
        pm(mean_stats[m]["ece"], std_stats[m]["ece"]),
        wins[m]["roc_auc"],
        wins[m]["accuracy"],
        wins[m]["f1"],
        wins[m]["log_loss"],
        wins[m]["ece"],
        pm(time_mean[m], time_std[m], digits=2),
    ])

summary_df = pd.DataFrame(rows, columns=pd.MultiIndex.from_tuples(col_tuples))
display(summary_df)

# Also provide raw numeric export-friendly version
raw_rows = []
for m in valid_models:
    raw_rows.append({
        "model": m,
        **{f"norm_{k}": norm_stats[m][k] for k in METRIC_COLS},
        **{f"norm_std_{k}": norm_std_stats[m][k] for k in METRIC_COLS},
        **{f"mean_{k}": mean_stats[m][k] for k in METRIC_COLS},
        **{f"std_{k}": std_stats[m][k] for k in METRIC_COLS},
        **{f"wins_{k}": wins[m][k] for k in METRIC_COLS},
        "mean_time_s": time_mean[m],
        "std_time_s": time_std[m],
    })

raw_summary_df = pd.DataFrame(raw_rows).sort_values("model").reset_index(drop=True)
raw_path = PROJECT_ROOT / f"binary_openml_summary_{timestamp_now()}.csv"
raw_summary_df.to_csv(raw_path, index=False)
print(f"Saved raw summary: {raw_path.resolve()}")
```
